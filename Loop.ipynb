```python
L=[1,2,3,4,4,"sudh","kite",'kumar',"abc", 'true','false',324,34.56,"abc"]
L1=[]
L2=[]
L3=[]
L4=[]
for i in L:
  if type (i)== int:
      L1.append(i)
  elif type(i)==float:
      L2.append(i)
  elif type(i)== bool:
      L3.append(i)
  else:
      L4.append(i)

print("list of num:",L1)
print("list of  float:" , L2)
print("list of boolean:",L3)
print("list of string:" , L4)
```

```python
n=int (input("enter a number"))
n2 =n
rev=0
while n>0:
     digit=n%10
     rev=(rev*10)+digit
     n=n//10
print(rev)
if rev==n2:
   print("n is palindrome")
else:
   print("n is not palindrome")
```
